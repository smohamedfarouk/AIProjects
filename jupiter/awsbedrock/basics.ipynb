```python
from langchain_aws import ChatBedrockConverse

llm = ChatBedrockConverse(
    model="nvidia.nemotron-super-3-120b",
    region_name="us-east-1",
    temperature=0.7,
    max_tokens=500
)

response = llm.invoke("Explain Agentic AI in simple words.")

print(response.content)
```

```python
from langchain_aws import ChatBedrockConverse

llm = ChatBedrockConverse(
    model="amazon.nova-lite-v1:0",
    region_name="us-east-1",
    disable_streaming=False,

    temperature=0.5,
    max_tokens=500
)

messages = [
    (
        "system",
        "You are an AI trainer. Explain concepts in simple language."
    ),
    (
        "human",
        "What is RAG?"
    )
]

response = llm.invoke(messages)

print(response.content)
```

```python
import boto3

client = boto3.client(
    service_name="bedrock-runtime",
    region_name="us-east-1"
)

response = client.converse(
    modelId="amazon.nova-lite-v1:0",

    messages=[
        {
            "role": "user",
            "content": [
                {
                    "text": "Explain Generative AI in simple words."
                }
            ]
        }
    ],

    inferenceConfig={
        "maxTokens": 500,
        "temperature": 0.7
    }
)

answer = response["output"]["message"]["content"][0]["text"]

print(answer)
```
