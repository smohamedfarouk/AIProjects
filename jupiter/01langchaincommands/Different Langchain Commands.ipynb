```python
from tavily import TavilyClient
```

```python
import os
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv())
tavily = TavilyClient(api_key=os.getenv("TAVILY_API_KEY"))
```

```python
def internet_search(query: str, max_results: int = 5):
    """
    Search the web for current information on topic. Use for any question that needs up-to date information, facts
    or trends or competitor/market information 
    """
    return tavily.search(query=query, max_results=max_results)
```

```python
from deepagents import create_deep_agent
```

```python
web_agent = create_deep_agent(
    tools = [internet_search],
    system_prompt= (
        "You are a web agent that can search the internet for information."
        "You are an expert Researcher, Always (1) Use Write-Todos to Plan"
        "(2) Use Internet Search Tool to gather Information, (3) Save findings to file"
        "(4) write a final report"
    ),
    model = llm
)
```

```python
out = web_agent.invoke({
    "messages": [{"role": "user", 
                   "content": "Latest News in AI Space as of Sep 16th 2026"}]
                   
})
```

```python
print(out['messages'][-1].content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
draft_prompt = ChatPromptTemplate.from_template(
    "Draft a short polite email to a client declining their Corporate Training Proposal due to Low Budget: {input}"
)
```

```python
draft_chain = draft_prompt | llm | StrOutputParser() #LCEL(Langchain Expression Language)
```

```python
print(draft_chain.invoke({"input": "The client offered 25000/day instead of 1lakh/day for AI Corporate Training"}))
```

```python
medical_prompt = ChatPromptTemplate.from_template(
    "Explain about this medicine Pros and Cons and side effects in 2Lines and End with Thanks for Reaching us out www.bepec.in: {Name}"
)
```

```python
critique_prompt = ChatPromptTemplate.from_template(
    "Identify any tone issues or rudeness in this email draft, if it is perfect, say no isses. \n Draft: {draft}"
)
```

```python
critique_chain = critique_prompt | llm | StrOutputParser() 
```

```python
refine_prompt = ChatPromptTemplate.from_template(
    "Rewrite the email based on the critique. \nOriginal Draft: {draft}\n Critique: {critique}\n Final Email:"
)
```

```python
refine_chain = refine_prompt | llm | StrOutputParser()
    
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
chain = (
    {"draft": draft_chain, "input": RunnablePassthrough()}
    | RunnablePassthrough.assign(critique = critique_chain)
    |refine_chain
)
```

```python
print(chain.invoke("The client offered 25000/day instead of 1lakh/day for AI Corporate Training"))
```
